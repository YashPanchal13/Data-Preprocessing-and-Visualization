```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer, Trainer, TrainingArguments
from datasets import Dataset
```

```python
import torch
import json

# Load the dataset from a JSON file
with open('/content/dataset/dataset20.json', 'r') as file:  # Replace 'dataset.json' with the path to your JSON file
    data = json.load(file)
```

```python
dataset = Dataset.from_dict({'text': [entry['user_prompt'] + '\n' + json.dumps(entry.get('poster_data', '')) for entry in data]})
# Load pre-trained GPT-2 model and tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
# Add padding token to the tokenizer
tokenizer.pad_token = tokenizer.eos_token # Set pad token to eos token
model = GPT2LMHeadModel.from_pretrained("gpt2", pad_token_id=tokenizer.eos_token_id)
# Tokenize the dataset
def tokenize_function(examples):
    tokenized = tokenizer(examples["text"], truncation=True, padding="max_length", max_length=512)
    tokenized['labels'] = tokenized['input_ids'].copy()  # Add labels for loss calculation

    # return tokenizer(examples["text"], truncation=True, padding="max_length", max_length=512)
    return tokenized

tokenized_datasets = dataset.map(tokenize_function, batched=True)

# Prepare training arguments
training_args = TrainingArguments(
    output_dir="./gpt2-poster-generation",  # Where to save the model
    num_train_epochs=3,
    per_device_train_batch_size=2,
    logging_dir="./logs",
    logging_steps=10,
    save_steps=10,
)

# Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets,
)

# Fine-tune the model
trainer.train()
```

```python
# Save the fine-tuned model
model.save_pretrained("./gpt2-poster-generation")
tokenizer.save_pretrained("./gpt2-poster-generation")
```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Load the fine-tuned model
# tokenizer = GPT2Tokenizer.from_pretrained("./gpt2-poster-generation")
# model = GPT2LMHeadModel.from_pretrained("./gpt2-poster-generation")

tokenizer = GPT2Tokenizer.from_pretrained("/content/gpt2-poster-generation")
model = GPT2LMHeadModel.from_pretrained("/content/gpt2-poster-generation")
# Function to generate poster data from a user prompt
def generate_poster_from_prompt(prompt):
    input_ids = tokenizer.encode(f"User Prompt: {prompt}", return_tensors="pt")
    output = model.generate(input_ids, max_length=150, num_beams=5, no_repeat_ngram_size=2, early_stopping=True)
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)

    return generated_text

# Example usage
user_prompt = input("Enter your prompt: ")
poster_data = generate_poster_from_prompt(user_prompt)
print(poster_data)

```
